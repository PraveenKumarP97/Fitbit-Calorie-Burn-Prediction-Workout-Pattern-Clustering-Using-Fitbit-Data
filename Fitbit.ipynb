```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
def evaluate_model(model_name, Y_test, pred):
    mae = mean_absolute_error(Y_test, pred)
    rmse = np.sqrt(mean_squared_error(Y_test, pred))
    r2 = r2_score(Y_test, pred)

    print(f"{model_name} Results")
    print("MAE:", mae)
    print("RMSE:", rmse)
    print("R2 Score:", r2)
    print("-"*50)
```

```python
from sklearn.model_selection import cross_val_score

def cross_validate_model(model, X_data, Y_data, model_name):
    
    scores = cross_val_score(
        model,
        X_data,
        Y_data,
        cv=5,
        scoring='r2'
    )
    
    print(f"{model_name} Cross Validation R2 Scores:", scores)
    print(f"Average CV R2 Score: {scores.mean()}")
    print("-"*50)
```

**Unsupervised Learning**

```python
df = pd.read_csv("fitbit_final_cleaned.csv")
```

```python
df["BMI"] = df["Weight (kg)"] / (df["Height (m)"] ** 2)

df["Base_MET"] = 7
df["HR_Intensity"] = df["Avg_BPM"] / df["Max_BPM"]
df["Effective_MET"] = df["Base_MET"] * df["HR_Intensity"]
```

```python
cluster_features = [
    "Age",
    "Weight (kg)",
    "Height (m)",
    "Max_BPM",
    "Avg_BPM",
    "Resting_BPM",
    "Session_Duration (hours)",
    "Fat_Percentage",
    "Water_Intake (liters)",
    "Workout_Frequency (days/week)",
    "Experience_Level",
    "BMI",
    "Base_MET",
    "HR_Intensity",
    "Effective_MET"
]

cluster_df = df[cluster_features].copy()
```

```python
#Scaling
from sklearn.preprocessing import StandardScaler

scaler_cluster = StandardScaler()
cluster_scaled = scaler_cluster.fit_transform(cluster_df)
```

```python
#PCA(Principal Component Analysis)
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
cluster_pca = pca.fit_transform(cluster_scaled)
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=4, random_state=42)
clusters = kmeans.fit_predict(cluster_pca)
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(cluster_pca, clusters)
print("Silhouette Score:", score)
```

```python
df["Cluster"] = clusters
```

```python
joblib.dump(scaler_cluster, "scaler_cluster.pkl")
joblib.dump(pca, "pca_model.pkl")
joblib.dump(kmeans, "kmeans_model.pkl")
```

```python
cluster_columns = [
    "Age",
    "Weight (kg)",
    "Height (m)",
    "Max_BPM",
    "Avg_BPM",
    "Resting_BPM",
    "Session_Duration (hours)",
    "Fat_Percentage",
    "Water_Intake (liters)",
    "Workout_Frequency (days/week)",
    "Experience_Level",
    "BMI",
    "Base_MET",
    "HR_Intensity",
    "Effective_MET"
]

joblib.dump(cluster_columns, "cluster_columns.pkl")
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

wcss = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(cluster_pca)   # IMPORTANT: use PCA output
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(2, 11), wcss, marker='o')
plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS (Inertia)")
plt.grid(True)
plt.show()
```

WCSS (Within Cluster Sum of Squares) which will measures how compact your clusters are. The curve starts flattening significantly on K=4. Final Clustering,

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Create PCA dataframe
pca_df = pd.DataFrame({
    "PC1": cluster_pca[:, 0],
    "PC2": cluster_pca[:, 1],
    "Cluster": clusters
})

plt.figure(figsize=(10, 7))

# Scatter plot (cleaner styling)
sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Cluster",
    palette="tab10",
    s=90,
    alpha=0.75,
    edgecolor="black"
)

# ---------------------------------------------------
# ADD CENTROIDS (IMPORTANT IMPROVEMENT)
# ---------------------------------------------------
centroids = pca_df.groupby("Cluster")[["PC1", "PC2"]].mean()

plt.scatter(
    centroids["PC1"],
    centroids["PC2"],
    s=250,
    c="black",
    marker="X",
    label="Centroids"
)

# ---------------------------------------------------
# FINAL STYLING
# ---------------------------------------------------
plt.title("Workout Pattern Clusters (PCA + KMeans)", fontsize=14, fontweight="bold")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.legend()
plt.grid(True, linestyle="--", alpha=0.3)

plt.show()
```

**DBSCAN (Density-Based Spatial Clustering)**

```python
from sklearn.cluster import DBSCAN
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define the two epsilon values to compare
eps_values = [0.3, 0.7]
```

```python
# Create a figure with 1 row and 2 columns
fig, axes = plt.subplots(1, 2, figsize=(18, 7), sharey=True)

for i, eps in enumerate(eps_values):
    # 1. Run DBSCAN with current epsilon
    dbscan = DBSCAN(eps=eps, min_samples=5)
    dbscan_clusters = dbscan.fit_predict(cluster_pca)
    
    # 2. Create temp DataFrame for plotting
    pca_df = pd.DataFrame({
        "PC1": cluster_pca[:, 0],
        "PC2": cluster_pca[:, 1],
        "Cluster": dbscan_clusters
    })
    
    # 3. Plot on the specific subplot (axes[i])
    sns.scatterplot(
        data=pca_df,
        x="PC1",
        y="PC2",
        hue="Cluster",
        palette="tab10",
        s=90,
        alpha=0.75,
        edgecolor="black",
        ax=axes[i]
    )
    
    # Calculate number of clusters (excluding noise -1)
    n_clusters = len(set(dbscan_clusters)) - (1 if -1 in dbscan_clusters else 0)
    
    # 4. Formatting each subplot
    axes[i].set_title(f"DBSCAN: eps={eps} | Clusters Found: {n_clusters}")
    axes[i].set_xlabel("Principal Component 1")
    if i == 0:
        axes[i].set_ylabel("Principal Component 2")
    axes[i].grid(True, linestyle="--", alpha=0.3)
    axes[i].legend(title="Cluster", loc='upper right')

plt.tight_layout()
plt.show()
```

This dataset was densly packed. DBSCAN was tested as an alternative clustering technique. However, it failed to form multiple meaningful clusters due to overlapping density distributions in fitness behavioral data. Therefore, KMeans was selected as the final clustering model because it achieved better segmentation performance with a silhouette score of 0.356.
